```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('test_scores.csv')
df.head()
```

```python
plt.scatter(x = df['math'],y =df['cs'],color = 'red', marker = '+', linewidth=5)
```

```python
def gradient_descent(x, y):
    m_curr = b_curr = 0
    learning_rate = 0.0002
    n = len(df['math'])
    plt.scatter(x,y,color = 'red', marker = '+', linewidth='5')
    for i in range (100000):
        y_predicted = m_curr * x + b_curr
        print(m_curr, b_curr, i)
        m_derivative = -(2/n)*sum(x*(y-y_predicted))
        y_derivative = -(2/n)*sum((y-y_predicted))
        m_curr = m_curr - learning_rate * m_derivative
        b_curr = b_curr - learning_rate * y_derivative
    plt.figure(figsize=(8, 5))
    plt.scatter(x, y, color='red', marker='+', linewidth=2, label='Data Points')
    plt.plot(x, m_curr * x + b_curr, color='green', linewidth=2, label='Best Fit Line')
    plt.xlabel('math')
    plt.ylabel('cs')
    plt.legend()
    plt.show()
        
```

```python
def predict_using_sklearn():
    r = LinearRegression()
    r.fit(df[['math']], df.cs)
    print("USING SKLEARN, THIS THE SLOP AND INTERCEPT WE HAVE FOUND")
    print(f'm = {r.coef_}, b = {r.intercept_}')
```

```python
x = df['math']
y = df['cs']
```

```python
gradient_descent(x, y)
predict_using_sklearn()
```
