```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
```

```python
digits = load_digits()
dir(digits) #exploring what this dataset contains
```

```python
digits.data[0] #it's an array, 8x8 image represented as array.
```

```python
plt.gray()
plt.matshow(digits.images[0])
```

```python
for i in range (5):
    plt.matshow(digits.images[i])
```

```python
digits.target[0:5] #tells image 0 is 0, image 1 is 1, image 2 is 2, image 3 is 3, image 4 is 4
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.predict(X_test)
```

```python
lr.score(X_test, y_test)
```

```python
plt.matshow(digits.images[67])
```

```python
digits.target[67]
```

```python
lr.predict([digits.data[67]])
```

```python
lr.predict(digits.data[0:1000])
```

```python
from sklearn.metrics import confusion_matrix
```

```python
y_predicted = lr.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_predicted)
cm
```

```python
import seaborn as sns
plt.figure(figsize =(10,6))
sns.heatmap(cm, annot=True)
plt.xlabel("Predicted")
plt.ylabel('Truth')
```
